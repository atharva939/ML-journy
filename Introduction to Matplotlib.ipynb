* when ploting somthing advanced use oo method

```python
over_50 = Heart_deasis[Heart_deasis["age"]> 50]
over_50.head()
```

```python
over_50.plot(kind='scatter',
            x='age',
            y='chol',
            c='target',
            figsize=(10, 6))
```

```python
#oo method
fix, ax = plt.subplots(figsize=(10,6))
over_50.plot(kind='scatter',
            x='age',
            y='chol',
            c='target',
            ax=ax);

##ax.set_xlim([45,100])
```

```python
fif, ax =plt.subplots(figsize=(10,6))

scatter = ax.scatter(x=over_50["age"],
                    y=over_50["chol"],
                    c=over_50["target"]);
ax.set(title="Heart disease",
       xlabel="Age",
       ylabel="cholestrol");

ax.legend(*scatter.legend_elements(),title="target")

ax.axhline(over_50["chol"].mean(),
          linestyle='--');
```

```python
over_50.head()
```

```python
over_50.plot(kind='scatter',
            x='age',
            y='thalach',
            c='target');
```

```python
fif, ax =plt.subplots(figsize=(10,6))

scatter = ax.scatter(x=over_50["age"],
                    y=over_50["thalach"],
                    c=over_50["target"]);
ax.set(title="Heart disease",
       xlabel="Age",
       ylabel="thalach");

ax.legend(*scatter.legend_elements(),title="target")

ax.axhline(over_50["thalach"].mean(),
          linestyle='--');
```

```python
fig, [ax0,ax1]=plt.subplots(nrows=2,
                           ncols=1,figsize=(10,10),sharex=True);

scatter=ax0.scatter(x=over_50['age'],
                    y=over_50['chol'],
                    c=over_50['target'])

ax0.set(title="Heart disease",
       ylabel="chol");

ax0.legend(*scatter.legend_elements(),title="target")

ax0.axhline(over_50["chol"].mean(),
          linestyle='--');

scatter=ax1.scatter(x=over_50['age'],
                    y=over_50['thalach'],
                    c=over_50['target'])

ax1.set(title="Heart disease",
       xlabel="Age",
       ylabel="thalach");

ax1.legend(*scatter.legend_elements(),title="target")

ax1.axhline(over_50["thalach"].mean(),
          linestyle='--');


fig.suptitle("heart disease analysis",fontsize=16);
```

```python
fig, [ax0,ax1]=plt.subplots(nrows=2,
                           ncols=1,figsize=(10,10),sharex=True);

scatter=ax0.scatter(x=over_50['age'],
                    y=over_50['chol'],
                    c=over_50['target'],cmap="winter")

ax0.set(title="Heart disease",
       ylabel="chol");

ax0.set_xlim([50,80])

ax0.legend(*scatter.legend_elements(),title="target")

ax0.axhline(over_50["chol"].mean(),
          linestyle='--');

scatter=ax1.scatter(x=over_50['age'],
                    y=over_50['thalach'],
                    c=over_50['target'],cmap="winter")

ax1.set(title="Heart disease",
       xlabel="Age",
       ylabel="thalach");
ax1.set_xlim([50,80])
ax1.set_ylim([60,200])

ax1.legend(*scatter.legend_elements(),title="target")

ax1.axhline(over_50["thalach"].mean(),
          linestyle='--');


fig.suptitle("heart disease analysis",fontsize=16);
```
